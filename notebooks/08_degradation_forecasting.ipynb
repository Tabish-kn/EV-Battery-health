# 08. Battery Degradation Forecasting & Kinetics

Simulates long-term capacity fade curves using Arrhenius thermal acceleration and square-root cycle aging models.

```python
import os
import sys
sys.path.append('..')

import matplotlib.pyplot as plt
from src.degradation import simulate_degradation_trajectory, calculate_arrhenius_acceleration
```

```python
traj_cool = simulate_degradation_trajectory(current_soh=95.0, current_cycle=200, fast_charge_ratio=0.1, avg_temperature_c=25.0)
traj_hot = simulate_degradation_trajectory(current_soh=95.0, current_cycle=200, fast_charge_ratio=0.7, avg_temperature_c=45.0)

plt.figure(figsize=(10, 5))
plt.plot(traj_cool['cycles'], traj_cool['projected_soh'], label='Gentle Use (25°C, 10% Fast Chg)', color='#10b981', lw=2)
plt.plot(traj_hot['cycles'], traj_hot['projected_soh'], label='Aggressive Stress (45°C, 70% Fast Chg)', color='#ef4444', lw=2)
plt.axhline(70.0, color='gray', linestyle='--', label='EOL Threshold (70% SOH)')
plt.title('Projected Battery SOH Degradation Trajectories')
plt.xlabel('Battery Equivalent Cycles')
plt.ylabel('State of Health (%)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Key Findings
- High temperature and heavy fast-charging drastically accelerate capacity fade.

### Limitations
- Semi-empirical model requires periodic recalibration with physical BMS readouts.

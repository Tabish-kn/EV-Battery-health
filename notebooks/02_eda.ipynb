# 02. Advanced Exploratory Data Analysis (EDA)

In-depth visualization and statistical correlation analysis of battery electrical, thermal, and usage telemetry.

```python
import os
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from src.data_loader import load_data
```

```python
df = load_data('../data/raw/ev_battery_health.csv')
print(f'Telemetry records loaded: {len(df):,}')
```

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
sns.histplot(df['battery_failure'], bins=2, color='#ef4444')
plt.title('Battery Failure Distribution')

plt.subplot(1, 3, 2)
sns.histplot(df['state_of_health'].dropna(), kde=True, color='#3b82f6')
plt.title('SOH Distribution (%)')

plt.subplot(1, 3, 3)
sns.histplot(df['predicted_remaining_life_cycles'].dropna(), kde=True, color='#10b981')
plt.title('RUL Cycles Distribution')
plt.tight_layout()
plt.show()
```

```python
num_cols = ['cell_temperature_max', 'internal_resistance', 'voltage_imbalance', 'capacity_loss_percent', 'thermal_health_score', 'state_of_health', 'predicted_remaining_life_cycles', 'battery_failure']
corr = df[num_cols].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', cbar=True)
plt.title('Correlation Heatmap with Battery Failure')
plt.show()
```

### Key Findings
- Failure rate is 6.92% (13.45:1 class imbalance).
- Maximum cell temperature and thermal runaway risk are strongly correlated with failure.
- State of Health (SOH) negatively correlates with capacity loss and internal resistance.

### Limitations
- Cross-sectional snapshots of pack diagnostics.

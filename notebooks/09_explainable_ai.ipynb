# 09. Explainable AI (XAI) & SHAP Interpretability

Computes global TreeSHAP feature attributions and local explanations for individual battery diagnostic predictions.

```python
import os
import sys
sys.path.append('..')

from src.models import ModelBundle
from src.data_loader import load_data
from src.explainability import BatteryModelExplainer
from src.feature_engine import get_feature_matrix
```

```python
failure_bundle = ModelBundle.load('../models/failure_model')
explainer = BatteryModelExplainer(failure_bundle)
df = load_data('../data/raw/ev_battery_health.csv')

X_sample, _, _ = get_feature_matrix(df.head(1), target_type='failure')
explanation = explainer.explain_instance(X_sample)
print('Top Risk Increasing Factors:')
for f in explanation['top_risk_increasing_factors']:
    print(f' + {f["feature"]}: {f["feature_value"]} (SHAP: {f["shap_value"]:+.4f})')
```

### Key Findings
- TreeSHAP clearly identifies thermal runaway risk and internal resistance as primary failure contributors.

### Limitations
- SHAP represents model attribution, not physical causation.

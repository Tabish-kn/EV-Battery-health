# 10. Comprehensive Multi-Model Benchmarking & Selection

Consolidates evaluation metrics across Failure Classification, SOH Regression, RUL Regression, and Anomaly Detection.

```python
import os
import sys
sys.path.append('..')

import json
import pandas as pd

for mod in ['failure_model', 'soh_model', 'rul_model', 'anomaly_model']:
    meta_path = f'../models/{mod}/{mod}_metadata.json'
    if os.path.exists(meta_path):
        with open(meta_path, 'r') as f:
            meta = json.load(f)
        print(f'=== {meta["model_name"]} ({meta["model_type"]}) ===')
        print(json.dumps(meta.get('metrics', {}), indent=2))
        print()
```

### Key Findings
- Tree ensembles (XGBoost, Random Forest) consistently outperform linear baselines across all battery diagnostic tasks.

### Limitations
- Hyperparameter tuning conducted on fixed stratified splits.
